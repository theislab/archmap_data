```python
import scanpy as sc
from scvi.hub import HubModel

sc.set_figure_params(figsize=(4, 4))

# for white background of figures (only for docs rendering)
%config InlineBackend.print_figure_kwargs={'facecolor' : "w"}
%config InlineBackend.figure_format='retina'
```

First we specify the revision, which is the commit hash, found [here](https://huggingface.co/scvi-tools/human-lung-cell-atlas-scanvi/commits/main). For more info on how to download a model from scVI-Hub, see the official tutorial [here](https://docs.scvi-tools.org/en/1.0.0/tutorials/notebooks/scvi_hub_intro_and_download.html). Please specify the directory that your downloaded files will save to. By default, all files will save to "scvihub_model/"

```python
download_dir = "scvihub_model/"

hmo = HubModel.pull_from_huggingface_hub(
    repo_name="scvi-tools/human-lung-cell-atlas-scanvi",
    cache_dir=download_dir,
    revision="cdc082c0e69d8794602332b1cdb2de747c113442",
)
```

```python
hmo.adata
```

You will find your downloaded models under the directory download_dir + "models--scvi-tools--human-lung-cell-atlas-scanvi/snapshots/cdc082c0e69d8794602332b1cdb2de747c113442". These files can now be uploaded to ArchMap.
