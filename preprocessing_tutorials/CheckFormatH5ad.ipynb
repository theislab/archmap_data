```python
import tempfile
import scanpy as sc

def check_data(data_path):
    # 1️⃣ Load data
    try:
        query_adata_raw = sc.read(data_path)
        print("Data successfully loaded.")
    except Exception as e:
        raise RuntimeError(
            f"Error message: {e}. There is likely an issue with the way your data "
            "(anndata object) is formatted upon upload. Please reach out to ArchMap "
            "(archmap.bio@gmail.com) with a screenshot of this error."
        )

    if "batch" not in query_adata_raw.obs.columns:
        raise ValueError(
            "Batch key not specified. Please ensure your batch key is labelled 'batch'."
        )

    if query_adata_raw.n_obs > 200000:
        raise ValueError(
            "The number of cells exceeds 200,000. Please split your data into batches."
        )

    try:
        temp_query = tempfile.NamedTemporaryFile(suffix=".h5ad")
        query_adata_raw.write_h5ad(temp_query.name)

        print("All checks completed. Your query conforms to all requirements :)")

    except ValueError as e:
        if "is also used by a column whose values are different" in str(e):
            raise ValueError(
                f"{e}. Please check for duplicate column names in .obs and .var."
            ) from e
        else:
            raise ValueError(
                f"{e}. There is likely an issue with your anndata formatting. "
                "Please contact ArchMap with this error."
            )



```

```python
check_data(data_path)
```
