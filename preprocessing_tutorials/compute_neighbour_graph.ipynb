This colab notebook calculates the the neighbourhood graph for the full reference and query mapping results. This is to prepare the data for visualization with cellxgene.

```python
path_to_file = "test_data/delorey_latent_rep.h5ad" #please change path to input .h5ad file (your mapping results downloaded from ArchMap. You can upload this file to the colab session by clicking on Files -> Upload to session storage on the left sidebar).
output_path = "delorey1_with_umap.h5ad" #please change file output name to your desired name.
```

```python
!pip install scanpy
!pip install h5py
```

```python
import scanpy as sc
import h5py
import gc
from anndata.experimental import read_elem
with h5py.File(path_to_file, 'r') as store:
  latent_rep = read_elem(store['obsm/latent_rep'])
  obs = read_elem(store['obs'])
  adata = sc.AnnData(latent_rep, obs)
```

```python
del latent_rep
del obs
gc.collect()
```

```python
sc.pp.neighbors(adata, n_neighbors=8)
sc.tl.umap(adata)
```

```python
#3. Unique var index identifier
adata.var_names_make_unique()

#4. Unique obs index identifier
adata.obs_names_make_unique()
```

```python
adata.write(output_path)
```
