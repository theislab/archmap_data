After identifying your column name, please set the variable "gene_col" to the name of the column with either your gene symbols or Ensembl IDs. Then run the cell below. The saved h5ad file can now be used in ArchMap.

```python
gene_col = "gene_names"
query_adata_raw.var['var_index_archmap'] = query_adata_raw.var[gene_col].copy()
query_adata_raw.var_names = query_adata_raw.var['var_index_archmap']
query_adata_raw.var.drop(columns=["var_index_archmap"], inplace=True)
query_adata_raw.write("test_reformated.h5ad")
```
